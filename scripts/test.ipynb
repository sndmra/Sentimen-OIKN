```python
# --- Install library dulu (kalau belum ada) ---
# pip install pandas openpyxl fitter matplotlib

import pandas as pd
from fitter import Fitter
import matplotlib.pyplot as plt

# 1. Import data dari Excel
# Ganti 'Loss' dengan nama kolom di file kamu
data = pd.read_excel("simteo.xlsx")
loss_data = data['loss_data'].dropna().values  

# pip install distfit
from distfit import distfit

dfit = distfit()
dfit.fit_transform(loss_data)
dfit.plot()

print(dfit.summary)
```

```python
from transformers import pipeline
import re

# ===== GANTI MODEL =====
pretrained_name = "newreyy/sentiment-analysis-distilbert-IKN"

# Load model & tokenizer
nlp = pipeline(
    "sentiment-analysis",
    model=pretrained_name,
    tokenizer=pretrained_name
)

print(nlp("IKN sangat bagus."))
print(nlp("IKN buruk seperti istana iblis."))
```
